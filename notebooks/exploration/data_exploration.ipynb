```python
# Select only numeric feature columns
numeric_cols = data.select_dtypes(include=np.number).columns

# Rows containing at least one missing value
nan_mask = data.isna().any(axis=1)

# Rows containing at least one +inf or -inf value
inf_mask = np.isinf(data[numeric_cols]).any(axis=1)

# Count affected rows
nan_rows = nan_mask.sum()
inf_rows = inf_mask.sum()

# Rows containing BOTH NaN and Inf
both_rows = (nan_mask & inf_mask).sum()

# Unique rows containing either NaN OR Inf
problem_mask = nan_mask | inf_mask
problem_rows = problem_mask.sum()

# Percentage of entire dataset affected
problem_percentage = (problem_rows / len(data)) * 100

print("Total rows:", len(data))
print("Rows with NaN:", nan_rows)
print("Rows with Inf:", inf_rows)
print("Rows with BOTH:", both_rows)
print("Unique problematic rows:", problem_rows)
print("Percentage affected:", problem_percentage, "%")
```

```python
invalid_structure_mask = (
    (df_clean["Fwd Header Length"] < 0) |
    (df_clean["Bwd Header Length"] < 0) |
    (df_clean["min_seg_size_forward"] < 0)
)

print("Removing:", invalid_structure_mask.sum())

df_clean = df_clean.loc[~invalid_structure_mask].copy()
df_clean.reset_index(drop=True, inplace=True)

print("New shape:", df_clean.shape)
```
