```python
import pandas as pd
import numpy as np

files = {
    "Monday": "../../data/raw/Monday-WorkingHours.pcap_ISCX.csv",
    "Tuesday": "../../data/raw/Tuesday-WorkingHours.pcap_ISCX.csv",
    "Wednesday": "../../data/raw/Wednesday-workingHours.pcap_ISCX.csv",
    "Thursday_AM": "../../data/raw/Thursday-WorkingHours-Morning-WebAttacks.pcap_ISCX.csv",
    "Thursday_PM": "../../data/raw/Thursday-WorkingHours-Afternoon-Infilteration.pcap_ISCX.csv",
    "Friday_AM": "../../data/raw/Friday-WorkingHours-Morning.pcap_ISCX.csv",
    "Friday_PortScan": "../../data/raw/Friday-WorkingHours-Afternoon-PortScan.pcap_ISCX.csv",
    "Friday_DDoS": "../../data/raw/Friday-WorkingHours-Afternoon-DDos.pcap_ISCX.csv"
}
```

```python
dfs = {}

for name, path in files.items():
    df = pd.read_csv(path, low_memory=False)
    dfs[name] = df
```

```python
'''
1. Dataset Schema Validation
'''
# Display shape and number of columns for every file

for name, df in dfs.items():
    print(f"{name:15} Rows: {df.shape[0]:,} | Columns: {df.shape[1]}")

reference_columns = list(dfs["Monday"].columns)

for name, df in dfs.items():
    same_schema = list(df.columns) == reference_columns
    print(f"{name:15} Same schema: {same_schema}")

for i, col in enumerate(dfs["Monday"].columns, start=1):
    print(f"{i:2}. {repr(col)}")
```

Combining all datasets

```python
for name, df in dfs.items():
    df["source_file"] = name

data = pd.concat(dfs.values(), ignore_index=True)

print(f"Rows: {data.shape[0]:,}")
print(f"Columns: {data.shape[1]}")
print(f"Classes: {data['Label'].nunique()}")
```

```python
data["source_file"].value_counts()
```

```python
class_distribution = (
    data["Label"]
    .value_counts()
    .rename_axis("Class")
    .reset_index(name="Count")
)

class_distribution["Percentage"] = (
    class_distribution["Count"] / len(data) * 100
)

class_distribution["Percentage"] = (
    class_distribution["Percentage"].round(4)
)

class_distribution
```

```python
missing_count = data.isnull().sum()

missing_report = pd.DataFrame({
    "Missing_Count": missing_count,
    "Missing_Percentage": (
        missing_count / len(data) * 100
    )
})

missing_report = (
    missing_report[
        missing_report["Missing_Count"] > 0
    ]
    .sort_values("Missing_Count", ascending=False)
)

missing_report
```

```python
numeric_cols = data.select_dtypes(include=np.number).columns

pos_inf = np.isposinf(data[numeric_cols]).sum()
neg_inf = np.isneginf(data[numeric_cols]).sum()

inf_report = pd.DataFrame({
    "Positive_Inf": pos_inf,
    "Negative_Inf": neg_inf
})

inf_report["Total_Inf"] = (
    inf_report["Positive_Inf"] +
    inf_report["Negative_Inf"]
)

inf_report = (
    inf_report[
        inf_report["Total_Inf"] > 0
    ]
    .sort_values("Total_Inf", ascending=False)
)

inf_report
```

```python
duplicates_with_source = data.duplicated().sum()

print(
    f"Exact duplicates including source_file: "
    f"{duplicates_with_source:,}"
)
```

```python
original_cols = [
    col for col in data.columns
    if col != "source_file"
]

duplicates_without_source = (
    data.duplicated(subset=original_cols).sum()
)

print(
    f"Duplicates ignoring source_file: "
    f"{duplicates_without_source:,}"
)
```

```python
duplicate_mask = data.duplicated(
    subset=original_cols,
    keep=False
)

data.loc[
    duplicate_mask,
    "Label"
].value_counts()
```

```python
feature_cols = [
    col for col in data.columns
    if col not in ["Label", "source_file"]
]

unique_counts = data[feature_cols].nunique(dropna=False)

constant_features = (
    unique_counts[unique_counts <= 1]
    .sort_values()
)

constant_features
```
