```python
nan_rows = data[data.isna().any(axis=1)] #atleast 1 NaN in row
inf_rows = data[data.isin([float('inf'), float('-inf')]).any(axis=1)]
both_rows = data[data.isna().any(axis=1) & data.isin([float('inf'), float('-inf')]).any(axis=1)]
either_rows = data[data.isna().any(axis=1) | data.isin([float('inf'), float('-inf')]).any(axis=1)]

print(nan_rows)
print(inf_rows)
print(both_rows)
print(either_rows)
```

```python
# Select only numeric feature columns
numeric_cols = data.select_dtypes(include=np.number).columns

# Rows containing at least one missing value
nan_mask = data.isna().any(axis=1)

# Rows containing at least one +inf or -inf value
inf_mask = np.isinf(data[numeric_cols]).any(axis=1)

# Count affected rows
nan_rows = nan_mask.sum()
inf_rows = inf_mask.sum()

# Rows containing BOTH NaN and Inf
both_rows = (nan_mask & inf_mask).sum()

# Unique rows containing either NaN OR Inf
problem_mask = nan_mask | inf_mask
problem_rows = problem_mask.sum()

# Percentage of entire dataset affected
problem_percentage = (problem_rows / len(data)) * 100

print("Total rows:", len(data))
print("Rows with NaN:", nan_rows)
print("Rows with Inf:", inf_rows)
print("Rows with BOTH:", both_rows)
print("Unique problematic rows:", problem_rows)
print("Percentage affected:", problem_percentage, "%")
```

```python
problem_mask = nan_mask | inf_mask
# Get only the problematic rows
problem_rows = data.loc[problem_mask]

# How many problematic rows belong to each class?
problem_by_class = problem_rows["Label"].value_counts()

# How many total rows belong to each class?
total_by_class = data["Label"].value_counts()

# Put them together
class_impact = pd.DataFrame({
    "Total Records": total_by_class,
    "Problematic Records": problem_by_class
}).fillna(0)

# Calculate what percentage of EACH CLASS is affected
class_impact["% of Class Affected"] = (
    class_impact["Problematic Records"]
    / class_impact["Total Records"]
    * 100
)

class_impact = class_impact.sort_values(
    "% of Class Affected",
    ascending=False
)

display(class_impact)
```

```python
feature_cols = [
    col for col in df_clean.columns
    if col not in ["Label", "source_file"]
]

feature_duplicates = df_clean.duplicated(
    subset=feature_cols
).sum()

print(
    "Duplicate feature vectors:",
    feature_duplicates
)

label_counts_per_pattern = (
    df_clean
    .groupby(feature_cols, dropna=False)["Label"]
    .nunique()
)

conflicting_patterns = label_counts_per_pattern[
    label_counts_per_pattern > 1
]

print(
    "Feature patterns with conflicting labels:",
    len(conflicting_patterns)
)
```

```python
conflict_keys = conflicting_patterns.index

conflict_rows = (
    df_clean
    .set_index(feature_cols)
    .loc[conflict_keys]
    .reset_index()
)

print("Conflicting feature patterns:", len(conflicting_patterns))
print("Total rows involved:", len(conflict_rows))

print("\nLabels involved:")
print(conflict_rows["Label"].value_counts())
```

```python
conflict_combinations = (
    conflict_rows
    .groupby(feature_cols)["Label"]
    .agg(lambda x: tuple(sorted(set(x))))
    .value_counts()
)

print(conflict_combinations)
```

```python
# Create a MultiIndex representing every row's feature vector
row_feature_index = pd.MultiIndex.from_frame(
    df_clean[feature_cols]
)

# conflicting_patterns.index contains the ambiguous feature vectors
conflict_mask = row_feature_index.isin(
    conflicting_patterns.index
)

print("Rows identified for conflict removal:", conflict_mask.sum())
```

```python
UNKNOWN_CLASS = "Bot"

unknown_df = df_clean[
    df_clean["Label"] == UNKNOWN_CLASS
].copy()

known_df = df_clean[
    df_clean["Label"] != UNKNOWN_CLASS
].copy()

print("Full clean dataset :", len(df_clean))
print("Known dataset      :", len(known_df))
print("Unknown Bot dataset:", len(unknown_df))
```

```python
from sklearn.model_selection import train_test_split

RANDOM_STATE = 42

train_df, temp_df = train_test_split(
    known_df,
    test_size=0.30,
    random_state=RANDOM_STATE,
    stratify=known_df["Label"]
)

val_df, test_df = train_test_split(
    temp_df,
    test_size=0.50,
    random_state=RANDOM_STATE,
    stratify=temp_df["Label"]
)
```
